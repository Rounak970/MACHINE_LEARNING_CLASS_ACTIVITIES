```python
import numpy as np
from sklearn.decomposition import PCA

# Define a 2x2 matrix
X = np.array([
    [2, 4],
    [6, 8]
])

print("Original Matrix:\n", X)

# Apply PCA
pca = PCA(n_components=2)

# Fit and transform
X_pca = pca.fit_transform(X)

print("\nPCA Transformed Data:\n", X_pca)
print("\nPrincipal Components:\n", pca.components_)
print("\nExplained Variance:\n", pca.explained_variance_)
```

```python
import numpy as np

# Define the same 2x2 matrix
X = np.array([
    [2, 4],
    [6, 8]
])

print("Original Matrix:\n", X)

# Apply SVD
U, S, Vt = np.linalg.svd(X)

print("\nU Matrix:\n", U)
print("\nSingular Values:\n", S)
print("\nV Transpose Matrix:\n", Vt)
```
